```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler
from sklearn.impute import SimpleImputer
from sklearn.metrics import classification_report, accuracy_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, StackingClassifier
from xgboost import XGBClassifier
```

```python
train_df = pd.read_csv("atlantis_citizens_final.csv")
test_df  = pd.read_csv("test_atlantis_hidden.csv")
```

```python
train_df = train_df.dropna(subset=["Wealth_Index", "House_Size_sq_ft", "Life_Expectancy"])
```

```python
train_df
```

```python
X_train = train_df[["Diet_Type","District_Name","Wealth_Index","House_Size_sq_ft","Life_Expectancy","Vehicle_Owned","Work_District"]]
X_train
```

```python
y_train = train_df["Occupation"].map({"Warrior":0, "Merchant":1, "Fisher":2, "Miner":3, "Scribe":4})
X_test = test_df[["Diet_Type","District_Name","Wealth_Index","House_Size_sq_ft","Life_Expectancy","Vehicle_Owned","Work_District"]]
X_test
```

```python
sim = SimpleImputer(strategy="median")
scaler = MinMaxScaler()
```

```python
list_one = ["Wealth_Index", "House_Size_sq_ft", "Life_Expectancy"]
# this line of code is not important since i have already dropped all the columns containing the na values

X_train.loc[:, list_one] = sim.fit_transform(X_train[list_one])
X_test.loc[:, list_one] = sim.transform(X_test[list_one])

X_train.loc[:, list_one] = scaler.fit_transform(X_train[list_one])
X_test.loc[:, list_one] = scaler.transform(X_test[list_one])
```

```python
l_two = ["Diet_Type", "District_Name", "Vehicle_Owned", "Work_District"]
```

```python
im = SimpleImputer(strategy="most_frequent")
X_train.loc[:, l_two] = im.fit_transform(X_train[l_two])
X_test.loc[:, l_two] = im.transform(X_test[l_two])

ohe = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
ohe.fit(X_train[l_two]) 
```

```python
tr_ohe = pd.DataFrame(ohe.transform(X_train[l_two]), columns=ohe.get_feature_names_out(l_two), index=X_train.index)
te_ohe = pd.DataFrame(ohe.transform(X_test[l_two]), columns=ohe.get_feature_names_out(l_two), index=X_test.index)
```

```python
X_trainnn = pd.concat([X_train[list_one], tr_ohe], axis=1)
X_testtt = pd.concat([X_test[list_one], te_ohe], axis=1)
```

```python
############################################################

xgbc = XGBClassifier(n_estimators=200,learning_rate=0.1,max_depth=3,subsample=0.6,colsample_bytree=1,eval_metric="mlogloss",num_class=5,random_state=42,n_jobs=-1)
```

```python
rfc = RandomForestClassifier(n_estimators=500,max_depth=20,max_features="sqrt",min_samples_leaf=2,min_samples_split=10,random_state=42,n_jobs=-1)
```

```python
lr = LogisticRegression(max_iter=3000,n_jobs=-1)
```

```python
sc = StackingClassifier( estimators=[("xgb", xgbc),("rf", rfc),("lr", lr),],stack_method="predict_proba",n_jobs=-1,)
```

```python
sc.fit(X_trainnn, y_train)
```

```python
test_solve = sc.predict(X_testtt)
```

```python
sub = pd.DataFrame({ "Citizen_ID": test_df["Citizen_ID"], "Occupation": test_solve})
```

```python
sub.to_csv("sub16.csv", index=False)
```
